```python
hospital["chronic_disease"] = hospital["chronic_disease"].fillna("No")
```

```python
hospital["total_cost"] = hospital["medicine_cost"] + hospital["test_cost"] + hospital["consultation_fee"]
```

```python
hospital["risk_score"] = (
    hospital["emergency_case"] +
    hospital["missed_previous_appointments"] +
    hospital["waiting_time_minutes"]/10
)
```

```python
hospital["waiting_category"] = pd.cut(
    hospital["waiting_time_minutes"],
    bins=[0, 15, 30, 60, 200],
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
hospital["satisfaction_level"] = pd.cut(
    hospital["patient_satisfaction_score"],
    bins=[0, 2, 3.5, 5],
    labels=["Low", "Medium", "High"]
)
```

```python
hospital["satisfaction_level"] = pd.cut(
    hospital["patient_satisfaction_score"],
    bins=[0, 5, 8, 10],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)
```

```python
hospital["department"].value_counts().plot(
    kind="bar",
    figsize=(8,4),
    title="Department-wise Patient Load",
    color="skyblue"
)
plt.ylabel("Number of Patients")
plt.show()
```

```python
sns.histplot(hospital["waiting_time_minutes"], bins=30, kde=True)
plt.title("Waiting Time Distribution")
plt.show()
```

```python
hospital["emergency_case"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    colors=["lightgreen", "red"]
)
plt.title("Emergency Case Distribution")
plt.ylabel("")
plt.show()
```

```python
sns.histplot(hospital["risk_score"], bins=30, kde=True)
plt.title("Patient Risk Score Distribution")
plt.show()
```

```python
hospital["satisfaction_level"].value_counts().plot(
    kind="bar",
    color=["green","orange","red"]
)
plt.title("Patient Satisfaction Levels")
plt.show()
```
